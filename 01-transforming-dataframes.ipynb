# Exploring DataFrames

When you get a new DataFrame, the first thing you need to do is explore it and see what it contains. Several useful methods and attributes can help with this:

- `.head()`: returns the first few rows of the DataFrame.

- `.info()`: shows information on each of the columns, such as the data type and number of missing values.

- `.shape`: returns the number of rows and columns of the DataFrame.

- `.describe()`: calculates a few summary statistics for each column. 

```python
import pandas as pd

homelessness = pd.DataFrame({
    "region": ["East South Atlantic", "Pacific", "Mountain", "New England", "South Atlantic"],
    "state": ["Alabama", "California", "Colorado", "Connecticut", "Florida"],
    "individuals": [2570.0, 109008.0, 7607.0, 2280.0, 21443.0],
    "family_members": [864.0, 20964.0, 3250.0, 1696.0, 9953.0],
    "state_pop": [4887681, 39538223, 5691287, 3571520, 21538187]
})

print('head() method')
print(homelessness.head())

print('\ninfo() method')
print(homelessness.info())

print('\nshape attribute')
print(homelessness.shape)

print('\ndescribe() method')
print(homelessness.describe())
```

# Components of a DataFrame

A DataFrame consists of three components, stored as attributes:

- `.values`: A two-dimensional NumPy array of values.

- `.columns`: An index of columns, representing the column names. 

- `.index`: An index for the rows,  which can be row numbers or row names. 

```python
import pandas as pd

homelessness = pd.DataFrame({
    "region": ["East South Atlantic", "Pacific", "Mountain", "New England", "South Atlantic"],
    "state": ["Alabama", "California", "Colorado", "Connecticut", "Florida"],
    "individuals": [2570.0, 109008.0, 7607.0, 2280.0, 21443.0],
    "family_members": [864.0, 20964.0, 3250.0, 1696.0, 9953.0],
    "state_pop": [4887681, 39538223, 5691287, 3571520, 21538187]
})

print(homelessness.values)

print(homelessness.columns) # Index(['region', 'state', 'individuals', 'family_members', 'state_pop'], dtype='object')

print(homelessness.index) # RangeIndex(start=0, stop=5, step=1)
```
