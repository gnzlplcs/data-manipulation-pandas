# Transforming DataFrames

## Exploring DataFrames

When you get a new DataFrame, the first thing you need to do is explore it and see what it contains. Several useful methods and attributes can help with this:

- `.head()`: returns the first few rows of the DataFrame.

- `.info()`: shows information on each of the columns, such as the data type and number of missing values.

- `.shape`: returns the number of rows and columns of the DataFrame.

- `.describe()`: calculates a few summary statistics for each column. 

```python
import pandas as pd

homelessness = pd.DataFrame({
    "region": ["East South Atlantic", "Pacific", "Mountain", "New England", "South Atlantic"],
    "state": ["Alabama", "California", "Colorado", "Connecticut", "Florida"],
    "individuals": [2570.0, 109008.0, 7607.0, 2280.0, 21443.0],
    "family_members": [864.0, 20964.0, 3250.0, 1696.0, 9953.0],
    "state_pop": [4887681, 39538223, 5691287, 3571520, 21538187]
})

print('head() method')
print(homelessness.head())

print('\ninfo() method')
print(homelessness.info())

print('\nshape attribute')
print(homelessness.shape)

print('\ndescribe() method')
print(homelessness.describe())
```

## Components of a DataFrame

A DataFrame consists of three components, stored as attributes:

- `.values`: A two-dimensional NumPy array of values.

- `.columns`: An index of columns, representing the column names. 

- `.index`: An index for the rows,  which can be row numbers or row names. 

```python
import pandas as pd

homelessness = pd.DataFrame({
    "region": ["East South Atlantic", "Pacific", "Mountain", "New England", "South Atlantic"],
    "state": ["Alabama", "California", "Colorado", "Connecticut", "Florida"],
    "individuals": [2570.0, 109008.0, 7607.0, 2280.0, 21443.0],
    "family_members": [864.0, 20964.0, 3250.0, 1696.0, 9953.0],
    "state_pop": [4887681, 39538223, 5691287, 3571520, 21538187]
})

print(homelessness.values)

print(homelessness.columns) # Index(['region', 'state', 'individuals', 'family_members', 'state_pop'], dtype='object')

print(homelessness.index) # RangeIndex(start=0, stop=5, step=1)
```

- `values`: Provides the actual data in a 2D NumPy array.

```python
import pandas as pd

homelessness = pd.DataFrame({
    "region": ["East South Atlantic", "Pacific", "Mountain", "New England", "South Atlantic"],
    "state": ["Alabama", "California", "Colorado", "Connecticut", "Florida"],
    "individuals": [2570.0, 109008.0, 7607.0, 2280.0, 21443.0],
    "family_members": [864.0, 20964.0, 3250.0, 1696.0, 9953.0],
    "state_pop": [4887681, 39538223, 5691287, 3571520, 21538187]
})

print(homelessness.values)
print(homelessness.columns)
print(homelessness.index)
```

## Sorting DataFrames

Sorting rows in a DataFrame can help you find the most interesting data quickly. You can use the `sort_values` method to sort by a specific column. For example, sorting by the `weight_kg` column will arrange the DataFrame from the lightest to the heaviest dog. 

```python
import pandas as pd

dogs = pd.DataFrame({
		'name': ['Stella', 'Bernie', 'Charlie', 'Lucy', 'Bella'],
		'weight_kg': [3, 70, 20, 20, 20],
		'height_cm': [20, 75, 30, 25, 35]
})

# Sorting by weight
sorted_dogs = dogs.sort_values('weight_kg')
print(sorted_dogs)
```

To sort by multiple columns, pass a list of columns names to `sort_values`.

```python
import pandas as pd

dogs = pd.DataFrame({
		'name': ['Stella', 'Bernie', 'Charlie', 'Lucy', 'Bella'],
		'weight_kg': [3, 70, 20, 20, 20],
		'height_cm': [20, 75, 30, 25, 35]
})

# Sorting by weight
sorted_dogs = dogs.sort_values(['weight_kg', 'height_cm'])
print(sorted_dogs)
```

## Subsetting Columns

When working with large datasets, you might onyly need a few columns. For example, to select only the `name` column from the `dogs` DataFrame:

```python
import pandas as pd

dogs = pd.DataFrame({
		'name': ['Stella', 'Bernie', 'Charlie', 'Lucy', 'Bella'],
		'weight_kg': [3, 70, 20, 20, 20],
		'height_cm': [20, 75, 30, 25, 35]
})

names = dogs['name']
print(names)

# To select multiple columns, use double square brackets.
name_weight = dogs[['name','weight_kg']]
print(name_weight)
```

## Subsetting Rows by Conditions

Subsetting rows based on conditions is a powerful way to filter data. Use relational operators to create logical conditions and filter rows. 

```python
import pandas as pd

dogs = pd.DataFrame({
		'name': ['Stella', 'Bernie', 'Charlie', 'Lucy', 'Bella'],
		'weight_kg': [3, 70, 20, 20, 20],
		'height_cm': [20, 75, 30, 25, 35],
		'color': ['beige', 'tan', 'white', 'black', 'brown']
})

# Subset rows where height in greater than 60 cm
tall_dogs = dogs[dogs['height_cm'] > 60]
print(tall_dogs.head())

# Subset rows where color is tan
tan_dogs = dogs[dogs['color'] == 'tan']
print(tan_dogs.head())

# Combine multiple conditions
tall_tan_dogs = dogs[(dogs['color'] == 'tan') & (dogs['height_cm'] > 60)]
print(tall_tan_dogs)
```
